```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

```python
# let's create a tool
@tool
def multiply(a : int, b : int)-> int:
  """Given 2 numbers a and b this tool returns their product"""
  return a * b
```

```python
print(multiply.invoke({'a': 5, 'b': 6}))
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
# tool binding
llm = ChatOpenAI()
```

```python
llm.invoke('hi')
```

```python
llm_with_tools = llm.bind_tools([multiply])
llm_with_tools.invoke('Hi how are you')
```

```python
# HumanMessage
query = HumanMessage('can you multiply 10 with 3')
messages = [query]
```

```python
messages
```

```python
result = llm_with_tools.invoke(messages)
result
```

```python
result.tool_calls[0]
```

```python
#  Appending AIMessage
messages.append(result)
```

```python
messages
```

```python
tool_result = multiply.invoke(result.tool_calls[0])
tool_result
```

```python
#  Appending ToolMessage
messages.append(tool_result)
messages
```

```python
llm_with_tools.invoke(messages).content
```

```python
# tool creation
from langchain_core.tools import InjectedToolArg
from typing import Annotated
from google.colab import userdata
# Get API key securely from Colab Secrets
api_key = userdata.get("EXCHANGE_RATE_API_KEY")

@tool
def get_convertion_factor(base_currency:str, target_currency:str)-> float:
  """ This function fetches the currency conversion factor between a given base currecy and a target currency """
  url = f"https://v6.exchangerate-api.com/v6/{api_key}/pair/{base_currency}/{target_currency}"
  response = requests.get(url)
  return response.json()
@tool
def calculator(base_currency_value: int, conversion_rate: Annotated[float, InjectedToolArg]) -> float:
  """
  given a currency converion rate this function calculates the target currency value from a given base currency value
  """
  return base_currency_value * conversion_rate
```

```python
calculator.args
```

```python
get_convertion_factor.invoke({'base_currency': 'USD','target_currency': 'INR'})
```

```python
calculator.invoke({'base_currency_value': 10, 'conversion_rate': 95.2528 })
```

```python
# tool binding
llm = ChatOpenAI()
llm_with_tools = llm.bind_tools([get_convertion_factor, calculator])
```

```python
messages = [HumanMessage('What is the conversion factor between INR and USD, and based on that can you convert 10 inr to usd')]
```

```python
messages
```

```python
ai_message = llm_with_tools.invoke(messages)
ai_message
```

```python
messages.append(ai_message)
```

```python
ai_message.tool_calls
```

```python
import json
for tool_call in ai_message.tool_calls:
  if tool_call['name'] == 'get_convertion_factor':
    tool_message1 = get_convertion_factor.invoke(tool_call)
    convertion_rate = json.loads(tool_message1.content)['conversion_rate']
    messages.append(tool_message1)
  if tool_call['name'] == 'calculator':
     tool_call['args']['conversion_rate'] = convertion_rate
     tool_message2 = calculator.invoke(tool_call)
     messages.append(tool_message2)
```

```python
messages
```

```python
llm_with_tools.invoke(messages)
```
