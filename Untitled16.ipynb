```python
class Person:
  pass
p = Person()
# p is object to the person class
print(p)
# Person → class
# Person() → creates/instantiates an object
# p → variable referring to that object
```

```python
class Manoj:
  pass
m = Manoj()
m.name = 'manoj'
m.age = 20
print(m.name)
# Here m is the object and name and age is the attribute
```

```python
class Manoj:
  def __init__(self, name, age):
    self.name = name
    self.age = age
m = Manoj('vinay', 20)
print(m.name)
print(m.age)
# self.name as creating an instance attribute or variable to the name parameter
# passing vinay and 20 as an argument to the constructor which goes into the parameter name, age.
# and in turn stores in attribute self.name
# Manoj → class
# m → object / instance
# "vinay" → argument
# name in def __init__(self, name) → parameter
# self → reference to the current instance or object m
# self.name → instance attribute
# self.name = name → assignment to an instance attribute
# __init__ → initializer method (commonly called the constructor in everyday Python discussion)
```

```python
def introduce(name, age):
  print(name)
  print(age)
person = {
    'name': 'Manoj',
    'age': 20
}
introduce(**person)
```
