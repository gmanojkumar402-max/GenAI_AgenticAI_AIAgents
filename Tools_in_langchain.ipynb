UsingStructuredTool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class Multiplyinput(BaseModel):
  a: int = Field(required = True, description = "the first number to add")
  b: int = Field(required = True, description = "the second number to be add")
```

```python
def multiply(a,b):
  return a*b
```

```python
tool = StructuredTool.from_function(
    func = multiply,
    name = 'multiply',
    description = 'multiply two numbers',
    args_schema = Multiplyinput
)
```

```python
result = tool.invoke({'a': 3, 'b': 4})
print(result)
```

```python
print(tool.name)
print(tool.description)
print(tool.args)
```

UsingBaseToolClass:

BaseTool is used to create custom LangChain tools with a standard interface. _run() defines synchronous execution, and _arun() can define asynchronous execution.

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic
class Multiplyinputs(BaseModel):
  a: int = Field(required = True, description = 'the first number to multiply')
  b: int = Field(required = True, description = 'the second number to multiply')
```

```python
class MultiplyTool(BaseTool):
  name:str  = 'multiply'
  description:str = 'multiply two numbers'
  args_schema: Type[BaseModel] = Multiplyinputs
  def _run(seld, a:int, b:int) -> int:
    return a*b
```

```python
tool1 = MultiplyTool()
result = tool1.invoke({'a': 5, 'b': 5})
print(result)
```

ToolKit

```python
from langchain_core.tools import tool
# Custom tools
@tool
def add(a: int, b: int) -> int:
  """Add two numbers"""
  return a+b

@tool
def multiply(a: int, b: int) -> int:
  """Multiply two numbers"""
  return a*b
```

```python
class ToolKit:
  def get_tools(self):
    return [add,multiply]
```

```python
tool5 = ToolKit()
tools = tool5.get_tools()
for tool in tools:
  print(tool.name)
  print(tool.description)
  print(tool.args)
```
