```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
import requests
from langchain_community.tools import DuckDuckGoSearchRun
```

Agent needs 2 things LLM's for text generation and reasoning and Tools to take action

search engine is our tool

```python
search_tool = DuckDuckGoSearchRun()
```

```python
# Read from Colab Secrets and put into environment
os.environ["WEATHERSTACK_API_KEY"] = userdata.get("WEATHERSTACK_API_KEY")
weather_api_key = os.getenv("WEATHERSTACK_API_KEY")
@tool
def get_weather(city:str) -> dict:
  """
  Fetch the current weather condition for the given city

  """
  api_key = f'https://api.weatherstack.com/current?access_key={weather_api_key}&query={city}'
  response = requests.get(api_key)
  return response.json()
```

Initializing llm

```python
llm = ChatOpenAI()
```

```python
from langchain.agents import create_agent
agent = create_agent(
    model = llm,
    tools = [search_tool, get_weather]
)
```

```python
response = agent.invoke({
    'messages': [
        {
            'role': 'user',
            'content': 'where is bothell located in and what is its current weather condition '
        }
    ]
})
```
